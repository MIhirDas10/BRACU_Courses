```python
import matplotlib.pyplot as plt
import numpy as np
import cv2
import os
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
folder_path = '/content/drive/MyDrive/463_LAB2'
image_files = []
for f in os.listdir(folder_path):
    if f.lower().endswith('.jpg'):
        img_path = os.path.join(folder_path, f)
        img = cv2.imread(img_path, cv2.IMREAD_COLOR)
        image_files.append(img)
print(len(image_files))
```

**step1**


`1`

```python
image1 = image_files[0]
clear_image = cv2.cvtColor(image1, cv2.COLOR_BGR2GRAY)
kernel = np.array([[0, 0, 0],
                   [0, 1, 0],
                   [0, 0, 0]])

conv = cv2.filter2D(clear_image, -1, kernel)
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 4, 1)
plt.imshow(clear_image, cmap='gray')
plt.title(f'Clear Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(conv, cmap='gray')
plt.title(f'After using kernel')
plt.axis('off')

plt.show()
```

For using 3*3 identity kernel, visually it doesn't really change anything.

Although, the size should reduce but cv2 uses padding default.

`3`

```python
image1 = image_files[2]
clear_image = cv2.cvtColor(image1, cv2.COLOR_BGR2RGB)
padding = ((3, 3), (3, 3), (0, 0))

zero = np.pad(clear_image, padding, mode='constant', constant_values=0)  # zero padding
zero = cv2.filter2D(zero, -1, kernel)

edge = np.pad(clear_image, padding, mode='edge') # edge padding
edge = cv2.filter2D(edge, -1, kernel)

reflect = np.pad(clear_image, padding, mode='reflect') # reflect padding
reflect = cv2.filter2D(reflect, -1, kernel)

plt.figure(figsize=(15, 6))
plt.subplot(1, 4, 1)
plt.title("Original Image")
plt.imshow(clear_image)
plt.axis('off')

plt.subplot(1, 4, 2)
plt.title("Zero Padding")
plt.imshow(zero)
plt.axis('off')

plt.subplot(1, 4, 3)
plt.title("Edge Padding")
plt.imshow(edge)
plt.axis('off')

plt.subplot(1, 4, 4)
plt.title("Reflect Padding")
plt.imshow(reflect)
plt.axis('off')

plt.show()
```

`4`

```python
def add_gaussian_noise(image, mean=0, sigma=25):
    noisy_image = image + np.random.normal(mean, sigma, image.shape).astype(np.uint8)
    return noisy_image

image_path = image_files[3]
clear_image = cv2.cvtColor(image_path, cv2.COLOR_BGR2RGB)

noisy_image = add_gaussian_noise(clear_image)
filtered_image = cv2.blur(noisy_image, (5, 5))
```

```python
plt.figure(figsize=(8, 4))
plt.subplot(1, 3, 1)
plt.imshow(clear_image)
plt.title('Clear Image')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(noisy_image)
plt.title('Noisy Image')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(filtered_image)
plt.title('Average Filter')
plt.axis('off')

plt.show()
```

reduces the amount of gaussian noise by averaging the pixel values

`5`

```python
clear_image = cv2.cvtColor(image_path, cv2.COLOR_BGR2RGB)

noisy_image = add_gaussian_noise(clear_image)
filtered_image = cv2.GaussianBlur(noisy_image, (5, 5), sigmaX=0)

plt.figure(figsize=(10, 6))
plt.subplot(1, 2, 1)
plt.imshow(noisy_image)
plt.title('Noisy Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(filtered_image)
plt.title('Filtered Image (Gaussian Filter)')
plt.axis('off')

plt.show()
```

```python
sigmaX = [10, 20, 30]
noisy_images = []
for i in sigmaX:
  noisy_image = cv2.GaussianBlur(noisy_image, (5, 5), i)
  noisy_images.append(noisy_image)

plt.figure(figsize=(10, 6))
plt.subplot(1, 3, 1)
plt.imshow(noisy_images[0])
plt.title(f'Noisy Image: sigma {sigmaX[0]}')
plt.axis('off')
plt.subplot(1, 3, 2)
plt.imshow(noisy_images[1])
plt.title(f'Noisy Image: sigma {sigmaX[1]}')
plt.axis('off')
plt.subplot(1, 3, 3)
plt.imshow(noisy_images[2])
plt.title(f'Noisy Image: sigma {sigmaX[2]}')
plt.axis('off')
plt.show()
```

smoother than the average filter according to the sigma value. For high value of sigma, blur becomes strong whereas for low value of sigma produces weaker blur. But more sigma value means more blur (smoothness), loss of details

```python
clear_image = cv2.cvtColor(image_files[4], cv2.COLOR_BGR2GRAY)

laplacian_filtered_image = cv2.Laplacian(clear_image, cv2.CV_64F)
laplacian_filtered_image = cv2.convertScaleAbs(laplacian_filtered_image)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(clear_image, cmap='gray')
plt.title('Clear Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(laplacian_filtered_image, cmap='gray')
plt.title('Laplacian Filter Applied')
plt.axis('off')

plt.show()
```

`7`

```python
clear_image = cv2.cvtColor(image_files[5], cv2.COLOR_BGR2GRAY)  # Convert to grayscale

def apply_convolution(img, kernel):
    return cv2.filter2D(img, -1, kernel)

vertical_edge_kernel = np.array([
                   [-1, 0, 1],
                   [-1, 0, 1],
                   [-1, 0, 1]])

horizontal_edge_kernel = np.array([
                   [-1, -1, -1],
                   [0, 0, 0],
                   [1, 1, 1]])

convolved_img_ver = apply_convolution(clear_image, vertical_edge_kernel)
convolved_img_hor = apply_convolution(clear_image, horizontal_edge_kernel)

plt.figure(figsize=(10, 5))
plt.subplot(1, 3, 1)
plt.title("Original Image")
plt.axis('off')
plt.imshow(clear_image, cmap='gray')

plt.subplot(1, 3, 2)
plt.title("Vertical Edge Kernel")
plt.imshow(convolved_img_ver, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title("Horizontal Edge Kernel")
plt.imshow(convolved_img_hor, cmap='gray')
plt.axis('off')
plt.show()
```

vertical edge kernel highlights left and right side wise pixel detection


---


horizontal edge kernel highlights top and bottom side wise pixel detection

```python
clear_image = cv2.cvtColor(image_files[6], cv2.COLOR_BGR2GRAY)
equalized_image = cv2.equalizeHist(clear_image)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(clear_image, cmap='gray')
plt.title('Original Grayscale Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(equalized_image, cmap='gray')
plt.title('Histogram Equalized Image')
plt.axis('off')
plt.show()
```

```python
first_equalization = cv2.equalizeHist(clear_image)
second_equalization = cv2.equalizeHist(first_equalization)
third_equalization = cv2.equalizeHist(second_equalization)

plt.figure(figsize=(15, 5))
plt.subplot(1, 4, 1)
plt.imshow(clear_image, cmap='gray')
plt.title(f'Clear Image')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(first_equalization, cmap='gray')
plt.title('1st Convolution')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(second_equalization, cmap='gray')
plt.title('2nd Convolution')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(third_equalization, cmap='gray')
plt.title(f'3rd Convolution')
plt.axis('off')

plt.show()
```

after the first histogram equalization from the clear image, the patterns become a lot clear but after that in the second, third one's don't show that much change (unnoticeable)
